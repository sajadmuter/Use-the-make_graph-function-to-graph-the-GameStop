# Apple Stock Data Exploration

This notebook uses AAPL (Apple), despite the current repository name referring to GameStop. It is a descriptive learning exercise. Saved outputs were cleared; live-source execution is pending.

```python
import json
import matplotlib.pyplot as plt
from analysis_utils import download_text, stock_history

CUTOFF = "2024-10-05"
```

## Company information sample

The IBM Skills Network JSON is a historical sample, not a source of current company facts.

```python
apple_info = json.loads(download_text("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/data/apple.json"))
print("Country in course sample:", apple_info["country"])
```

## Opening-price history

The end date is fixed, and auto_adjust=False is explicit. This does not predict share prices.

```python
apple_share_price_data = stock_history("AAPL", CUTOFF, price_column="Open")
print(apple_share_price_data.head())
ax = apple_share_price_data.plot(x="Date", y="Open", figsize=(11, 4), title="Apple historical opening price")
ax.set_ylabel("USD")
plt.tight_layout()
plt.show()
```
